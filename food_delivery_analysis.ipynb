# Food Delivery Data Analysis
This notebook contains analysis for the food delivery dataset.

```

import pandas as pd
import json
import sqlite3

orders = pd.read_csv("orders.csv")
with open("users.json") as f:
    users = pd.DataFrame(json.load(f))

conn = sqlite3.connect(":memory:")
conn.executescript(open("restaurants.sql").read())
restaurants = pd.read_sql("SELECT * FROM restaurants", conn)

df = orders.merge(users, on="user_id").merge(restaurants, on="restaurant_id")
df.head()
```
